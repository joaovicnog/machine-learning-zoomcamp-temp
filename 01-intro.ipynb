## Q6. Median value of horsepower

```python
median_before = df['horsepower'].median()
most_frequent = df['horsepower'].mode()[0]
df['horsepower'] = df['horsepower'].fillna(most_frequent)
median_after = df['horsepower'].median()

print("Has it changed?")
if median_before == median_after:
    print("\nNo")
elif median_after > median_before:
    print("\nYes, it increased")
else:
    print("\nYes, it decreased")
```

## Q7. Sum of weights

```python
X = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7).to_numpy()
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y
sum_of_weights = w.sum()
sum_of_weights
```
